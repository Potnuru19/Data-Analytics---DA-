```python
#Clustered Chart
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "house_value_category" : ["affordable", "moderate", "expensive", "premium", "luxury"],
    "population" : [1015, 1129, 333, 515, 624],
    "total_rooms" : [5612, 7650, 720, 1501, 1454]
})

df.plot(
    x="house_value_category",
    y=["population", "total_rooms"],
    kind="bar",
    width=0.8
)
plt.title("population and total rooms")
plt.xlabel("house value category")
plt.ylabel("count")
plt.legend()
plt.show()
```

```python
#Stacked Chart
house_value_category = ["affordable", "moderate", "expensive", "premium", "luxury"] #Stacked Chart
population = [1015, 1129, 333, 515, 624]
households = [472, 463, 117, 226, 262]
plt.bar(house_value_category, population, label="Population",color="green")
plt.bar(house_value_category, households, bottom=population, label="Households",color = "red")

plt.title("Population and Households")
plt.xlabel("House Value Category")
plt.ylabel("Count")

plt.legend()

plt.show()
```

```python
#Area Chart
house_value_category = ["affordable", "moderate", "expensive", "premium", "luxury"]
population = [1015, 1129, 333, 515, 624]
plt.plot(house_value_category, population,color="red")
plt.fill_between(house_value_category, population, color="red")
plt.title("Housing")
plt.xlabel("house value category")
plt.ylabel("population count")
plt.show()
```

```python
#Pie Chart
plt.pie(
    population,
    labels=house_value_category,
    autopct="%2.1f%%"
)

plt.title("Housing Population Distribution")
plt.show()
```

```python
#Scatter
housing_age = [15, 19, 17, 14, 20]
median_house_value = [66900, 80100, 85700, 73400, 65500]
plt.scatter(housing_age, median_house_value)
plt.title("Housing Age vs Median House Value")
plt.xlabel("Housing Age")
plt.ylabel("Median House Value")
plt.show()
```

```python
#Donut Chart
house_value_category = ["affordable", "moderate", "expensive", "premium", "luxury"] #donut chart
median_house_value = [66900, 80100, 85700, 73400, 65500]

plt.pie(
    median_house_value,
    labels=house_value_category,
    autopct="%1.1f%%",
    wedgeprops={"width":0.6}
)

plt.text(0, 0,
         "HouseValue",
         ha = "center",
         va = "center"
)
plt.show()
```

```python
#Boxplot
plt.boxplot(
    [housing_age, total_rooms, total_bedrooms],
    labels=["Housing Age", "Total Rooms", "Total Bedrooms"])
plt.title("Housing Data Analysis")
plt.ylabel("Values")
plt.show()
```
