```python
import matplotlib.pyplot as plt
import numpy as np
products = ['Laptop', "Mobile", "Tablet", "Headphones", "Watch"]
sales = [120, 180, 90, 150, 110]
profit = [30, 50, 20, 40, 25]
plt.plot(products, sales,marker="*")
plt.title("Product vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(products, sales)
plt.title("Product vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.barh(products, sales)
plt.title("Product vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(products, sales,label = "sales")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#multiple line chart
#clustered line chart
plt.plot(products, sales, marker="o", label="Sales")
plt.plot(products, profit, marker="*", label="Profit")
plt.title(" multiple line chart/Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "Products":["Laptop", "Mobile", "Tablet", "Monitor"], #Clustered Chart
    "Sales":[500, 700, 400, 600],
    "Profit":[100, 150, 80, 120]
})

df.plot(
    x="Products",
    y=["Sales", "Profit"],
    kind="bar",
    width=0.8
)
plt.title("Sales and Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
products=["A","B","C","D","E"] #Stacked Chart
sales=[200,300,400,250,350]
profit=[100,150,200,120,150]
plt.bar(products, sales, label="Sales")
plt.bar(products, profit, bottom=sales, label="Profit")

plt.title("Sales and Profit")
plt.xlabel("Products")
plt.ylabel("Amount")

plt.legend()

plt.show()
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"] #Area Chart
monthly_sales = [100, 120, 150, 140, 180, 200]

plt.plot(months, monthly_sales)
plt.fill_between(months, monthly_sales)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.show()
```

```python
plt.pie(
    sales,
    labels=products,
    autopct="%2.1f%%"
)

plt.title("Product Sales Distribution")
plt.show()
```
