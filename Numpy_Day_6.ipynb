```python
#horizontal split
arr = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80]
])
result = np.split(arr, 2, axis=1)
print(result)
```

```python
arr = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80]
])
result = np.split(arr, 2, axis=0)
print(result)
```
