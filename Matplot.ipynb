```python
import matplotlib.pyplot as plt
import numpy as np
products = ['Laptop', "Mobile", "Tablet", "Headphones", "Watch"]
sales = [120, 180, 90, 150, 110]
profit = [30, 50, 20, 40, 25]
plt.plot(products, sales,marker="*")
plt.title("Product vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(products, sales)
plt.title("Product vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.barh(products, sales)
plt.title("Product vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.show()
```

```python
plt.bar(products, sales,label = "sales")
plt.title("Product Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#multiple line chart
#clustered line chart
plt.plot(products, sales, marker="o", label="Sales")
plt.plot(products, profit, marker="*", label="Profit")
plt.title(" multiple line chart/Sales vs Profit")
plt.xlabel("Products")
plt.ylabel("Amount")
plt.legend()
plt.show()
```

```python
import matplotlib.pyplot as plt
months = ['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
sales  = [25000, 30000, 28000, 35000, 40000, 38000, 45000, 42000, 50000, 48000, 55000, 60000]
profit = [5000, 7000, 6000, 8000, 10000, 9000, 12000, 11000, 14000, 13000, 15000, 18000]
expenses = [20000, 23000, 22000, 27000, 30000, 29000, 33000, 31000, 36000, 35000, 40000, 42000]
```

```python
#1
plt.bar(months, sales,color="green")
plt.title("Sales vs Months")
plt.xlabel("Months") #Category
plt.ylabel("Sales")  #measure
plt.show()
```

```python
#2
plt.plot(months, sales,marker="*",color="black")
plt.title("Months vs Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#3
plt.plot(months, sales, marker="o", label="Sales")
plt.plot(months, profit, marker="*", label="Profit")
plt.title(" multiple line chart/Sales vs Profit")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

```python
#4
#month and expenses
plt.bar(months, expenses,color="red")
plt.title("Months vs Expenses")
plt.xlabel("Months") #Category
plt.ylabel("Expenses")  #measure
plt.show()
```

```python
#5
#The HR department wants to display the monthly sales performance in a way that makes it easy to compare each month's performance.
```

```python
#6
plt.plot(months, sales,marker="*",color="violet")
plt.title("Months vs Sales")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.show()
```

```python
#7
plt.plot(months, sales, marker="o", label="Sales",color="red")
plt.plot(months, expenses, marker="*", label="Expenses",color="yellow")
plt.plot(months, profit, marker="*", label="Profit",color="blue")
plt.title("Sales vs Expenses vs Profit")
plt.xlabel("Months")
plt.ylabel("Sales")
plt.legend()
plt.show()
```
